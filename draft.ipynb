```python
import os
import pandas as pd
from scipy.stats import spearmanr

# 你的文件路径
srcdir = "E:/SJTU/实习/国泰海通/业绩回测/result"
file_path = f"{srcdir}/中证500指增产品各收益回测/中证500指增_fund_ret_backtest.xlsx"

# 读取Excel所有sheet
xl_file = pd.ExcelFile(file_path)

# 遍历每个sheet，计算相关系数
result_list = []

for sheet_name in xl_file.sheet_names:
    print(f"正在处理 sheet: {sheet_name}")
    
    # 读取当前sheet
    df = pd.read_excel(file_path, sheet_name=sheet_name, index_col=0)
    
    # 确保列存在（防止报错）
    required_cols = ['年化收益', '年化波动率', 'ExcessExposre']
    if not all(col in df.columns for col in required_cols):
        print(f"⚠️ {sheet_name} 缺少必要列，跳过")
        continue
    
    # 去除空值（关键！否则相关系数算不出来）
    temp_df = df[required_cols].dropna()
    
    # 计算 年化收益 & ExcessExposure 斯皮尔曼
    corr_ret, p_ret = spearmanr(temp_df['年化收益'], temp_df['ExcessExposre'])
    
    # 计算 年化波动率 & ExcessExposre 斯皮尔曼
    corr_vol, p_vol = spearmanr(temp_df['年化波动率'], temp_df['ExcessExposre'])
    
    # 保存结果
    result_list.append({
        'sheet名称': sheet_name,
        '年化收益_ExcessExposure_相关系数': round(corr_ret, 4),
        '年化收益_p值': round(p_ret, 6),
        '年化波动率_ExcessExposure_相关系数': round(corr_vol, 4),
        '年化波动率_p值': round(p_vol, 6),
        '有效样本量': len(temp_df)
    })

# 输出最终结果表格
result_df = pd.DataFrame(result_list)
print("\n" + "="*80)
print("✅ 所有sheet 斯皮尔曼相关系数计算完成")
print("="*80)
print(result_df.to_string(index=False))

# 可选：自动保存结果到Excel
result_df.to_excel(f"{srcdir}/斯皮尔曼相关系数结果.xlsx", index=False)
print(f"\n📄 结果已保存到: {srcdir}/斯皮尔曼相关系数结果.xlsx")
```
